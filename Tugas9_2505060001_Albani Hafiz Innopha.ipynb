
**A. Membangun Katalog (Create)** *(bobot 20%)*

**B. Kueri Bertingkat (Read)** *(bobot 20%)*

Tampilkan hasil dari **tiga** kueri berikut :
1. Seluruh menu kategori "Kopi" dengan `tingkat_kepahitan` di atas 3.
2. Seluruh menu dengan `harga` di antara 15000 dan 30000.
3. Menu yang **tidak memiliki** field `pedas_level` sama sekali.

```python
# 1. Seluruh menu kategori "Kopi" dengan tingkat_kepahitan > 3
print("=== 1. Menu Kopi dengan Tingkat Kepahitan > 3 ===")
kueri_1 = koleksi_menu.find({"kategori": "Kopi", "tingkat_kepahitan": {"$gt": 3}})
for m in kueri_1:
    print(f"  - {m['nama_menu']} | Harga: Rp{m['harga']:,} | Kepahitan: {m['tingkat_kepahitan']}")

# 2. Seluruh menu dengan harga di antara 15000 dan 30000
print("\n=== 2. Menu dengan Harga antara 15.000 dan 30.000 ===")
kueri_2 = koleksi_menu.find({"harga": {"$gte": 15000, "$lte": 30000}})
for m in kueri_2:
    print(f"  - {m['nama_menu']} ({m['kategori']}) | Rp{m['harga']:,}")

# 3. Menu yang tidak memiliki field 'pedas_level'
print("\n=== 3. Menu Tanpa Field 'pedas_level' ===")
kueri_3 = koleksi_menu.find({"pedas_level": {"$exists": False}})
for m in kueri_3:
    print(f"  - {m['nama_menu']} | Kategori: {m['kategori']}")
```


**C. Pembaruan Data (Update)** *(bobot 15%)*

Tambahkan field `status` bernilai `"Tersedia"` ke seluruh dokumen menggunakan `update_many()`. Kemudian, ubah `status` menjadi `"Habis"` khusus untuk 2 menu pilihan anda menggunakan `update_one()`.

```python
# 1. Tambahkan field status="Tersedia" ke SELURUH dokumen
koleksi_menu.update_many({}, {"$set": {"status": "Tersedia"}})
print("Field status 'Tersedia' berhasil ditambahkan ke seluruh dokumen.")

# 2. Ubah status menjadi "Habis" untuk 2 menu pilihan
koleksi_menu.update_one({"nama_menu": "V60 Gayo"}, {"$set": {"status": "Habis"}})
koleksi_menu.update_one({"nama_menu": "Croissant Cokelat"}, {"$set": {"status": "Habis"}})
print("Status untuk 'V60 Gayo' dan 'Croissant Cokelat' diperbarui menjadi 'Habis'.")

# Verifikasi hasil pembaruan
print("\n--- Daftar Menu & Status Terbaru ---")
for m in koleksi_menu.find({}, {"_id": 0, "nama_menu": 1, "kategori": 1, "status": 1}):
    print(f"  - {m['nama_menu']} ({m['kategori']}): {m['status']}")
```


**D. Integrasi ke Spark** *(bobot 25%)*

Ambil seluruh data dari collection `menu`, ubah menjadi Spark DataFrame (ikuti pola Sub-bab 9.6 — perhatikan field yang mungkin kosong/`null` pada sebagian dokumen, seperti `tingkat_kepahitan` yang hanya dimiliki menu Kopi). Tampilkan:
1. `printSchema()` hasil konversi.
2. Rata-rata `harga` per `kategori` menggunakan `groupBy`.


```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import avg, count
import pandas as pd

# Inisialisasi SparkSession
spark = SparkSession.builder \
    .appName("Tugas9-KopiNusantara") \
    .master("local[*]") \
    .getOrCreate()
spark.sparkContext.setLogLevel("ERROR")

# 1. Ambil data dari MongoDB sebagai list of dict
dokumen_menu = list(koleksi_menu.find({}))

# 2. Ubah ke pandas DataFrame
df_pandas = pd.DataFrame(dokumen_menu)

# Penyesuaian tipe data objek khusus MongoDB & list
df_pandas["_id"] = df_pandas["_id"].astype(str)
if "notes_rasa" in df_pandas.columns:
    df_pandas["notes_rasa"] = df_pandas["notes_rasa"].apply(lambda x: ", ".join(x) if isinstance(x, list) else "")

# 3. Konversi ke Spark DataFrame
df_spark_menu = spark.createDataFrame(df_pandas)

# Tampilkan printSchema()
print("=== 1. Schema Spark DataFrame ===")
df_spark_menu.printSchema()

# Tampilkan Rata-rata harga per kategori menggunakan groupBy
print("\n=== 2. Rata-rata Harga per Kategori ===")
df_spark_menu.groupBy("kategori").agg(
    avg("harga").alias("rata_rata_harga"),
    count("nama_menu").alias("jumlah_menu")
).show()

spark.stop()
```


**E. Kesimpulan** *(bobot 20%)*

Tulis pada markdown cell (**minimal 100 kata**): apa tantangan yang anda temui saat mengonversi data MongoDB (dengan field yang tidak seragam antar dokumen) menjadi Spark DataFrame (yang mengharuskan skema tabel seragam)? Bagaimana anda mengatasinya?

Tantangan utama saat mengonversi data dari MongoDB ke Spark DataFrame terletak pada sifat benturan paradigma skema (schema mismatch). MongoDB mengusung prinsip schemaless atau flexible schema, di mana setiap dokumen dalam satu collection bisa memiliki kombinasi field yang berbeda, tipe data bertingkat (seperti array atau nested document), atau field opsional yang hanya ada pada kategori tertentu (misalnya field tingkat_kepahitan pada kopi dan pedas_level pada makanan). Sebaliknya, Spark DataFrame mewajibkan struktur skema tabel yang rigid dan seragam (strict schema) dengan tipe data yang terdefinisi secara konsisten di seluruh baris.

Saat dilakukan konversi melalui perantara Pandas (PyMongo -> Pandas -> Spark), field-field yang tidak dimiliki oleh suatu dokumen otomatis diisi nilai NaN atau None (berubah menjadi null di Spark). Selain itu, tipe data kompleks seperti BSON ObjectId dan array Python (seperti pada tags atau notes_rasa) dapat menyebabkan kegagalan inferensi tipe data pada Spark. Solusi yang dilakukan adalah melakukan sanitasi data awal (data cleaning) pada layer Pandas, yaitu mengonversi ObjectId menjadi tipe string, mengubah bentuk array/list menjadi teks terpisah koma, serta membiarkan Spark memetakan field yang tidak homogen sebagai kolom bernilai null agar agregasi seperti groupBy tetap dapat berjalan secara optimal tanpa error.
